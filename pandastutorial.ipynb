### Concat & Merge

```python
df2_data = {
    "dept": ["HR", "IT", "Finance"],
    "location": ["New York", "San Francisco", "Chicago"],
    "manager": ["Laura", "Steve", "David"],
}
```

```python
df2 = pd.DataFrame(df2_data)
```

```python
print(pd.concat([df, df2], axis=0))  # Row-wise concatenation
```

```python
print(pd.concat([df, df2], axis=1))  # Column-wise concatenation [01:00:33]
```

```python
merged_df = pd.merge(df, df2, on="dept")
```
